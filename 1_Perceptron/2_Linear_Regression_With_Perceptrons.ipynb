### Using Scikit-learn

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Perceptron
```

```python
from pathlib import Path

home_path = Path.home()
input_file_path = home_path.joinpath('GitHub/unlocking-deep-learning/98_Datasets/placements.csv')

df_placements = pd.read_csv(input_file_path)
df_placements.head()
```

```python
# Split the DataFrame into features (X) and label (y).
df_features = df_placements[["cgpa", "resume_score"]]
df_labels = df_placements["placed"]
```

```python
# first lets plot the input features in a graph
# taking 'do_the_movie_have_good_reviews' in x-axis and 'are_tickets_available' in y-axis
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

plt.scatter(df_placements["cgpa"], df_placements["resume_score"], c=df_labels, cmap='coolwarm', s=75, edgecolor='k')
plt.title('CGPA vs Resume Score by Placement Status')
plt.show()
```

```python
model = Perceptron()
model.fit(df_features, df_labels)
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(df_features.values, df_labels.values, clf=model, legend=2)
```
