### Imports

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Set random seed for reproducibility
torch.manual_seed(42)
```

### 1. Data Ingestion

### 2. Data Preparation

```python
## Here data ingestion and preparation are done in a single step for simplicity

# Sample data: square footage (in sq ft) and corresponding house prices (in USD)
sq_ft = torch.tensor([[1500.0], [2000.0], [2500.0], [3000.0], [3500.0]], dtype=torch.float32)
prices = torch.tensor([[295000.0], [405000.0], [490000.0], [610000.0], [695000.0]], dtype=torch.float32)
```

### 3. Modeling

```python
# create the model
model = nn.Sequential(
    nn.Linear(in_features=1, out_features=1)
    )
```

```python
# Define the loss function and optimizer
loss_fn = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.0000000001)
```

### 4. Training

```python
# Training loop
for epoch in range(500):
    # Zero gradients before backward pass
    optimizer.zero_grad()
    # Forward pass: compute predicted y by passing x to the model
    predictions = model(sq_ft)
    # Calculate the loss
    loss = loss_fn(predictions, prices)
    # perform a backward pass
    loss.backward()
    # update the weights
    optimizer.step()
    # Print loss every 50 epochs
    if (epoch + 1) % 50 == 0:
        print(f'Epoch {epoch + 1}: Loss= {loss.item()}')
```
